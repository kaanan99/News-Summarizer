```python
import os
import requests

from GoogleNews import GoogleNews
from sentence_transformers import SentenceTransformer, util
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
api = os.getenv("API")
```

```python
response = requests.get(f"{api}/topics")
```

```python
if response.status_code == 200:
    topic_data = dict(response.json())["data"]
else:
        print("Failed to retrieve content. Status code:", response.status_code)
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
model.max_seq_length = 128
```

```python
def get_headline_groups(model, headlines):
    embeddings = model.encode(headlines)
    clusters = util.community_detection(embeddings, min_community_size=3, threshold=0.6)
    return clusters

def get_headlines(topic_key):
    googlenews = GoogleNews(period="1d")
    googlenews.set_topic(topic_key)
    googlenews.get_news()
    news = googlenews.result()
    headlines = [obj["title"] for obj in news]

    return headlines
```

```python
for data in topic_data:
    topic_key = data["topic_key"]
    headlines = get_headlines(topic_key)
    clusters = get_headline_groups(model, headlines)
    print(headlines)
```
